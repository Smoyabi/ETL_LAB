```python
import pandas as pd
import sqlite3
import os

# Ensure 'loaded' folder exists
os.makedirs('loaded', exist_ok=True)
```

```python
# Load the transformed data
df_full = pd.read_csv('transformed/transformed_full.csv')
df_incremental = pd.read_csv('transformed/transformed_incremental.csv')

# Display preview
print("Full Data:")
display(df_full.head())

print("\nIncremental Data:")
display(df_incremental.head())
```

```python
# Connect to SQLite database
conn_full = sqlite3.connect('loaded/full_data.db')
conn_inc = sqlite3.connect('loaded/incremental_data.db')

# Load dataframes into SQLite tables
df_full.to_sql('full_data', conn_full, if_exists='replace', index=False)
df_incremental.to_sql('incremental_data', conn_inc, if_exists='replace', index=False)

# Verify load
print("Full Data from SQLite:")
display(pd.read_sql('SELECT * FROM full_data LIMIT 5', conn_full))

print("\nIncremental Data from SQLite:")
display(pd.read_sql('SELECT * FROM incremental_data LIMIT 5', conn_inc))

# Close connections
conn_full.close()
conn_inc.close()
```
